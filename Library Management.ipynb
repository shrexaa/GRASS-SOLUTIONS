```python
import json
import os
```

```python
def load_data(filename):
    if os.path.exists(filename):
        with open(filename, "r") as file:
            return json.load(file)
    return []

def save_data(filename, data):
    with open(filename, "w") as file:
        json.dump(data, file, indent=4)

books = load_data("books.json")
members = load_data("members.json")
issues = load_data("issues.json")
```

```python
def add_book():
    book_id = input("Enter book ID: ")

    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter book title: ")
    author = input("Enter author: ")

    try:
        copies = int(input("Enter number of copies: "))
        if copies <= 0:
            print("Copies must be greater than zero.")
            return
    except ValueError:
        print("Enter a valid number.")
        return

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "total": copies,
        "available": copies
    }

    books.append(book)
    save_data("books.json", books)

    print("Book added successfully!")
```

```python
def view_books():
    if len(books) == 0:
        print("No books available.")
        return

    for book in books:
        print("--------------------")
        print("ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Available:", book["available"])
        print("Total copies:", book["total"])
```

```python
def search_book():
    title = input("Enter book title to search: ").lower()

    found = False

    for book in books:
        if title in book["title"].lower():
            print("Book found!")
            print(book)
            found = True

    if not found:
        print("Book not found.")
```

```python
def add_member():
    member_id = input("Enter member ID: ")

    for member in members:
        if member["id"] == member_id:
            print("Member ID already exists!")
            return

    name = input("Enter member name: ")

    member = {
        "id": member_id,
        "name": name
    }

    members.append(member)
    save_data("members.json", members)

    print("Member registered successfully!")
```

```python
def issue_book():
    book_id = input("Enter book ID: ")
    member_id = input("Enter member ID: ")

    selected_book = None
    selected_member = None

    for book in books:
        if book["id"] == book_id:
            selected_book = book

    for member in members:
        if member["id"] == member_id:
            selected_member = member

    if selected_book is None:
        print("Book not found.")
        return

    if selected_member is None:
        print("Member not found.")
        return

    if selected_book["available"] == 0:
        print("No copies available.")
        return

    selected_book["available"] -= 1

    issue = {
        "book_id": book_id,
        "member_id": member_id,
        "returned": False
    }

    issues.append(issue)

    save_data("books.json", books)
    save_data("issues.json", issues)

    print("Book issued to", selected_member["name"])
```

```python
def return_book():
    book_id = input("Enter book ID: ")
    member_id = input("Enter member ID: ")

    for issue in issues:
        if (issue["book_id"] == book_id
                and issue["member_id"] == member_id
                and issue["returned"] == False):

            issue["returned"] = True

            for book in books:
                if book["id"] == book_id:
                    book["available"] += 1

            save_data("books.json", books)
            save_data("issues.json", issues)

            print("Book returned successfully!")
            return

    print("No matching active issue found.")
```

```python
def view_issued_books():
    found = False

    for issue in issues:
        if issue["returned"] == False:
            for book in books:
                if book["id"] == issue["book_id"]:
                    for member in members:
                        if member["id"] == issue["member_id"]:
                            print(
                                "Book:", book["title"],
                                "| Member:", member["name"]
                            )
                            found = True

    if not found:
        print("No books are currently issued.")
```

```python
while True:
    print("\n===== LIBRARY MANAGEMENT SYSTEM =====")
    print("1. Add Book")
    print("2. View Books")
    print("3. Search Book")
    print("4. Register Member")
    print("5. Issue Book")
    print("6. Return Book")
    print("7. View Issued Books")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        view_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        add_member()

    elif choice == "5":
        issue_book()

    elif choice == "6":
        return_book()

    elif choice == "7":
        view_issued_books()

    elif choice == "8":
        print("Library system closed.")
        break

    else:
        print("Invalid choice. Try again.")
```
